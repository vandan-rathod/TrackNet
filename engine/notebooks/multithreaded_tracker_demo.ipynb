# Multi-threaded video tracker

This notebook reads video frames in a background thread, leaving the main thread free for detection, OCR, and visualization.

```python
from pathlib import Path
import sys

PROJECT_ROOT = next(
    candidate for candidate in (Path.cwd().resolve(), *Path.cwd().resolve().parents)
    if (candidate / 'engine' / 'core' / 'tracker.py').exists()
)
sys.path.insert(0, str(PROJECT_ROOT))

from engine.core.tracker import MultiThreadingTracker
```

```python
# Use 0 for the default camera, or provide a path/stream URL.
SOURCE = 0

tracker = MultiThreadingTracker(queue_size=32).start(SOURCE)
try:
    while True:
        ok, frame = tracker.read(timeout=1)
        if not ok:
            if tracker.stopped:
                break
            continue

        # Run your detector/OCR pipeline here.
        print(frame.shape)
finally:
    tracker.release()

print(f'Frames captured: {tracker.frames_read}')
print(f'Capture error: {tracker.last_error}')
```
