```python
import cv2 as cv
import os
```

```python

# 1. FIX THE PATH: Use a raw string 'r' so Python ignores backslash escapes
folder_path = r"Dataset\Indian_Number_Plates\Sample_Images"

for img_path in os.listdir(folder_path):
    
    if img_path.endswith((".jpg",".jpeg",".png")):
        img=os.path.join(folder_path,img_path)
        img=cv.imread(img)

# 2. CHECK FIRST: Ensure the image is valid before doing ANY matrix operations
    if img is not None:
        h,w=img.shape[:2]

        max_height=600
        if h>max_height:
            scaleRatio=max_height/float(h)
            targetDimen=(int(w*scaleRatio),max_height)
            # 3. NOW it is safe to resize
            resized_frame = cv.resize(img, targetDimen, interpolation=cv.INTER_AREA)

        cv.imshow("window", resized_frame)
        cv.waitKey(0)
        cv.destroyAllWindows()
    else:
        print(f"Error: Could not load image. Looked at path: {folder_path}")
        print("Please double check if the folder names or file name matches exactly.")
```
